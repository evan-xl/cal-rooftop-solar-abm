# Exploration of LBNL Tracking the Sun Data

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = r'C:\Users\evanl\OneDrive\Desktop\Cal Classes\Honors Thesis\data\raw data\raw installations data\TTS_LBNL_public_file_29-Sep-2025_all.csv'
```

```python
tts = pd.read_csv(DATA_PATH)
tts.head()
```

```python
intermed_tts = tts.copy()
intermed_tts['installation_date'] = pd.to_datetime(intermed_tts.loc[:, 'installation_date'])
print('Time range:', intermed_tts['installation_date'].min(), 'to', intermed_tts['installation_date'].max())
```

```python
plt.hist(intermed_tts['installation_date'])
plt.xlabel('Time')
plt.ylabel('Count')
plt.title('Histogram of Installation Dates');
```

```python
plt.hist(intermed_tts[intermed_tts['state']=='CA']['installation_date'])
plt.xlabel('Time')
plt.ylabel('Count')
plt.title('Histogram of California Installation Dates');
```
